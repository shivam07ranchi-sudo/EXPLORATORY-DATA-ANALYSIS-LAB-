```python
import pandas as pd
import numpy as np

df=pd.read_csv('/1000 Sales Records.csv')
print(df.to_string())
print("-----------------------------------------------")
df.head() # for getting starting rows of the dataset and to get the number(n)of the rows just add n in the syntax that is (df.head(n))
print("------------------------------------------------")
df.tail(10)# for getting the ending rows of the data set and to get the numbers of row just add the n which the nu,mber of the rows you want in the syntax(df.tail(n))
print("-----------------------------------------------")
df.sample(4) # to get random or different n number of the rows
print("-----------------------------------------------")
df.shape# to the number of rows and the columns that is (x=rows ,y=columns)
print("-----------------------------------------------")
df.columns# tell all the columns name
print("-----------------------------------------------")
df.dtypes # it is used to get the data types of all columns
print("-----------------------------------------------")
df.info() # to get all the information about the used dataset
print("-----------------------------------------------")
df.describe()# to get count average mean ,min,max and (quartiles (25%,50%,75%) can get at once )
print("-----------------------------------------------")
df['Region'].unique()# used to get the all the unique values of the column of the dataset
print("-----------------------------------------------")
df['Region'].nunique()# used to get the count of the unique values in a mentiojed column of the dataset
print("-----------------------------------------------")
for col in df.columns:
    print(f"\n{col}")
    print("Number of unique values:", df[col].nunique())   #print values of all column unique count
    print(df[col].unique())
print("-----------------------------------------------")
for col in df.columns:
  print(f"\n{col}")
  print(f"number of the times {col} came in the dataset:")# to know how many times a value in the colum came in the dataset
  print(df[col].value_counts())
print("-----------------------------------------------")
df.index #row labels
print("-----------------------------------------------")

df.isnull().sum()# to get to know where and how many valu7es are missing or are null in the dataset
print("-----------------------------------------------")
df.isna().any()# to check whethere any column has missing value true or false
print("-----------------------------------------------")
df.dropna()# to remove the rows with the missing value
print("-----------------------------------------------")
df.dropna(subset=['Region'])#remove the row on the specific one value of a column
print("-----------------------------------------------")
# to fill any missing value by mean or median df['Region'].fillna(df['Region'].mean())
# at the place od mean put median for median
print("-----------------------------------------------")
# to drop a complete row or column
#dropna(axis=1)   column drop
#df.dropna(thresh=7)  means the row with less than 7 null values delete
print("-----------------------------------------------")
df.duplicated().sum()# to get total number of duplicate rows
print("-----------------------------------------------")
df[df.duplicated()]#to see duplicates rows
print("-----------------------------------------------")
df.drop_duplicates()#drop the duplicate rows
print("-----------------------------------------------")
df.drop_duplicates(subset=['Region']) # duplicate check only on the basis of the region colum
df.drop_duplicates(keep='last') #keep last  occurrence ,and remove the all first
df.sort_values('Units Sold',ascending=False).head(3)
df.sort_index()






```
